## 1. What is LangGraph, and why is it used for building AI agent workflows?

## What is LangGraph?

LangGraph is a Python framework used to build AI agent workflows. It helps manage the flow of tasks between multiple AI agents, maintain state, and handle decisions and loops.

### Why is LangGraph Used?

- **Connect Multiple AI Agents:** Coordinate multiple agents to complete complex tasks.
- **Maintain State:** Store and update workflow information between steps.
- **Conditional Routing:** Decide which agent or task should run next.
- **Handle Loops and Errors:** Repeat tasks when needed and manage failures.
- **Human Approval:** Allow humans to review and approve actions.
- **Build Reliable Applications:** Create complex AI applications with controlled workflows.

## 2.  What is the difference between LangGraph and LangChain?

## Difference Between LangChain and LangGraph

### LangChain
LangChain is a Python framework used to build LLM-powered applications. It helps connect language models with prompts, tools, databases, and external APIs.

### LangGraph
LangGraph is a Python framework used to build stateful AI agent workflows. It helps manage workflow steps, shared state, conditional routing, loops, and multiple agents.

### Key Differences

| Feature | LangChain | LangGraph |
|---|---|---|
| Purpose | Builds LLM-powered applications. | Builds complex AI agent workflows. |
| Main Components | Prompts, models, chains, and tools. | Nodes, edges, graphs, and state management. |
| Workflow | Suitable for straightforward workflows. | Suitable for workflows involving decisions and loops. |
| Processing | Supports sequential processing. | Supports sequential processing, conditional routing, and cycles. |
| Execution Control | Connects components to build applications. | Controls task execution and workflow progression. |
| State Management | Can manage state using its components and integrations. | Provides explicit shared state management across workflow steps. |

### Example

- **LangChain:** Send a question to an LLM and get an answer.
- **LangGraph:** A researcher agent collects information, an analyst agent analyzes it, and a writer agent generates the final report.

### In Simple Words

**LangChain** helps you build LLM applications, while **LangGraph** helps you control and manage complex AI agent workflows.

## 3. What is a `StateGraph` in LangGraph?

## What is StateGraph in LangGraph?

StateGraph in LangGraph is used to define an AI workflow by connecting different nodes and edges. It manages the shared state that stores and passes data between nodes.

### Main Components

- **State:** Stores workflow data and shares it between nodes.
- **Nodes:** Functions that perform tasks, such as calling an LLM or a tool.
- **Edges:** Define how the workflow moves from one node to another.
- **Conditional Edges:** Decide which node should execute next based on a condition.

### Example: AI Research Workflow

Suppose you build an AI research workflow.

1. **Researcher Node:** Collects information.
2. **Analyst Node:** Analyzes the information.
3. **Writer Node:** Generates the final report.

The StateGraph controls the execution order and passes the shared state between these nodes.

### In Simple Words

StateGraph is a structure that defines how tasks are connected, how data is shared, and how an AI workflow moves from one step to another.

## 4.  What is state in LangGraph, and how is it managed throughout a workflow?

## What is State in LangGraph?

State in LangGraph is a shared data structure that stores information used and updated by different nodes during a workflow.

### How Is State Managed?

1. **Initialize State:** Define the data fields required for the workflow, such as messages, research data, or results.
2. **Pass State:** LangGraph provides the current state to a node when it executes.
3. **Update State:** The node returns new data or updates existing fields.
4. **Merge Updates:** LangGraph applies the returned updates to the existing state according to the state schema and any configured reducers.
5. **Continue Workflow:** The updated state is passed to the next node.
6. **Persist State:** Checkpointing can save workflow state so execution can resume later.

### Example

In a research workflow:

- **Researcher Node:** Stores collected information in `research_data`.
- **Analyst Node:** Reads `research_data` and stores its analysis in `analysis`.
- **Writer Node:** Uses the research data and analysis to generate the final report.

### In Simple Words

State is like a shared notebook. Each node reads the information it needs and adds or updates information for the next node to use.

## 6.What are nodes and edges in LangGraph?

### Nodes

Nodes are Python functions that perform specific tasks in a LangGraph workflow. For example, calling an LLM, searching for information, or analyzing data.

### Edges

Edges define the flow between nodes. They determine which node executes next.

There are two main types of edges:

- **Normal Edges:** Move execution from one node to a specific next node.
- **Conditional Edges:** Choose the next node based on a condition or routing function.

### Example: AI Research Workflow

Suppose you build an AI research workflow:

1. **Researcher Node:** Collects information.
2. **Analyst Node:** Analyzes the information.
3. **Writer Node:** Generates the final report.

The edges connect these nodes in the required execution order.

### In Simple Words

**Nodes do the work, and edges decide where the workflow goes next.**

## 7.What is the difference between normal edges and conditional edges?

### 1. Normal Edges

Normal edges move the workflow from one node to a predefined next node. The next node does not depend on a routing condition.

### 2. Conditional Edges

Conditional edges decide which node executes next based on a condition or the output of a routing function.

### In Simple Words

- **Normal Edges:** Follow a fixed path.
- **Conditional Edges:** Choose the next path based on a condition.

````python
# 1. Normal Edge
graph.add_edge("researcher", "analyst")
````

````python
# 2. Conditional Edge
def router(state):
    if state["research_data"]:
        return "analyst"
    return "researcher"

graph.add_conditional_edges(
    "supervisor",
    router,
    {
        "analyst": "analyst",
        "researcher": "researcher"
    }
)
````

Here, the router checks the state and decides which node should execute next.

### Key Difference

- **Normal Edge:** Follows a predefined path.
- **Conditional Edge:** Chooses a path based on a condition.

### In Simple Words

A normal edge always follows a fixed route, while a conditional edge chooses the route based on the situation.

## 9. How does `add_conditional_edges()` work?

`add_conditional_edges()` is used in LangGraph to decide which node should execute next based on a condition or the output of a routing function.

It helps an AI workflow choose different paths dynamically.

### How It Works

1. **Node Execution:** A node finishes execution.
2. **Routing Function:** LangGraph calls the routing function with the current state.
3. **Choose a Path:** The routing function returns a value indicating the next path.
4. **Execute Next Node:** LangGraph executes the corresponding node.

### In Simple Words

`add_conditional_edges()` allows LangGraph to choose the next node dynamically based on the current state or a condition.


````python
from langgraph.graph import StateGraph, START, END

def router(state):
    if state["score"] >= 70:
        return "pass"
    return "fail"

graph = StateGraph(AgentState)

graph.add_node("check_result", check_result)
graph.add_node("pass_node", pass_node)
graph.add_node("fail_node", fail_node)

graph.add_edge(START, "check_result")

graph.add_conditional_edges(
    "check_result",
    router,
    {
        "pass": "pass_node",
        "fail": "fail_node"
    }
)

graph.add_edge("pass_node", END)
graph.add_edge("fail_node", END)
````

## 10.  What are reducers in LangGraph, and why are they important?

Reducers in LangGraph define how updates from nodes are combined with the existing state values.

### Why Are Reducers Important?

- **Control State Updates:** Define how new updates are merged with existing state values.
- **Combine Data:** Help combine data from multiple nodes.
- **Prevent Overwriting:** Avoid replacing existing data when accumulation is required.
- **Support Multiple Nodes:** Allow multiple nodes to contribute to the same state field.

### In Simple Words

Reducers define how new data is combined with existing data in the workflow state instead of simply overwriting it.

````python
#For example, LangGraph provides add_messages to manage message updates in the state.

from typing import TypedDict
from langgraph.graph import add_messages
from langgraph.graph.message import AnyMessage
from typing import Annotated

class AgentState(TypedDict):
    messages: Annotated[list[AnyMessage], add_messages]
````

## How Does `add_messages` Work in LangGraph?

- **Add New Messages:** If a node returns a new message, `add_messages` adds it to the existing message list.
- **Preserve History:** It preserves the conversation history.
- **Update Existing Messages:** If a message has the same ID as an existing message, it updates that message instead of simply appending a duplicate.

### Example

**Existing Messages:**

`[HumanMessage("Hi")]`

**New Update:**

`[AIMessage("Hello!")]`

**Updated Messages:**

`[HumanMessage("Hi"), AIMessage("Hello!")]`

### In Simple Words

`add_messages` is a reducer that manages conversation messages by adding new messages and updating existing messages when their IDs match.

## 11.  What is the purpose of `Annotated` when defining state fields with reducers?

`Annotated` is used to attach extra information to a state field, such as which reducer function LangGraph should use to process updates.

In this example, `Annotated` tells LangGraph to use `add_messages` as the reducer for the `messages` state field.


````python
class AgentState(TypedDict):
    messages: Annotated[list[AnyMessage], add_messages]
````

### What Does Each Part Mean?

- **`AgentState`:** Defines the structure of the LangGraph state.
- **`messages`:** The state field that stores messages.
- **`list[AnyMessage]`:** Specifies that the field contains a list of messages.
- **`Annotated`:** Attaches additional information to the field.
- **`add_messages`:** The reducer that tells LangGraph how to process message updates.

### Example of How It Works

Suppose the existing state contains one message:

`HumanMessage("Hi")`

A node returns a new message:

`AIMessage("Hello!")`

The `add_messages` reducer combines them, so the state contains both messages:

1. `HumanMessage("Hi")`
2. `AIMessage("Hello!")`

If a new message has the same ID as an existing message, `add_messages` updates that message instead of simply appending it.

### In Simple Words

`Annotated` connects the `messages` field to the `add_messages` reducer, which manages how messages are added or updated in the state.

## 13. How do you pass state updates from one node to another?


In LangGraph, state updates are passed from one node to another through the shared state. Each node receives the current state as input and returns a dictionary containing the fields it wants to update.

### How It Works

1. **Receive State:** The first node receives the current state as input.
2. **Return Updates:** The node returns updated values for specific fields.
3. **Apply Updates:** LangGraph applies those updates to the existing state.
4. **Pass Updated State:** The next node receives the updated state.

### In Simple Words

Each node reads the current state, returns its updates, and LangGraph applies those updates before passing the state to the next node.

````python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

class AgentState(TypedDict):
    question: str
    research_data: str

def researcher(state: AgentState):
    return {"research_data": "LangGraph manages AI workflows."}

def analyst(state: AgentState):
    data = state["research_data"]
    return {"research_data": data + " It uses nodes and edges."}

graph = StateGraph(AgentState)

graph.add_node("researcher", researcher)
graph.add_node("analyst", analyst)

graph.add_edge(START, "researcher")
graph.add_edge("researcher", "analyst")
graph.add_edge("analyst", END)

app = graph.compile()

result = app.invoke({
    "question": "What is LangGraph?",
    "research_data": ""
})

print(result["research_data"])
````

### Output

`LangGraph manages AI workflows. It uses nodes and edges.`

### What Happens Here?

- **Researcher Node:** Updates `research_data`.
- **LangGraph:** Applies the update to the shared state.
- **Analyst Node:** Receives the updated state and adds more information.
- **Final State:** Contains the combined result from the nodes.

### In Simple Words

A node returns a dictionary of updates, and LangGraph applies those updates to the shared state so the next node can access them.

# 14. What happens when multiple nodes update the same state field?


When multiple nodes update the same state field in LangGraph, the result depends on whether a reducer is defined for that field.

### 1. Without a Reducer

The new value replaces the existing value.

If multiple nodes in the same execution step update the same field, LangGraph may raise an `InvalidUpdateError` because it cannot automatically choose which value to keep.

### In Simple Words

Without a reducer, a new value replaces the existing value. If multiple nodes update the same field simultaneously, LangGraph may raise an error because it does not know how to combine those updates.

````python
# Without Reducer
class AgentState(TypedDict):
    result: str

def node_a(state):
    return {"result": "Research result"}

def node_b(state):
    return {"result": "Analysis result"}

# If both nodes update result in the same step, LangGraph raises an error because no reducer is defined to combine the updates.
````

````python
# with reducer
from typing import TypedDict, Annotated
import operator

class AgentState(TypedDict):
    results: Annotated[list, operator.add]

def node_a(state):
    return {"results": ["Research result"]}

def node_b(state):
    return {"results": ["Analysis result"]}
````


If both nodes run in the same step, their updates are combined into a list containing both results.

`operator.add` is a built-in Python function that performs addition for numbers and concatenation for lists or strings.

In LangGraph, it can act as a reducer to combine existing list values with new list updates instead of replacing them.

### In Simple Words

Without a reducer, conflicting updates to the same field can cause an error. With a reducer, LangGraph knows how to combine those updates.

## 15. What is the difference between sequential and parallel execution in LangGraph?


## Sequential vs Parallel Execution in LangGraph

### 1. Sequential Execution

Nodes execute one after another. The next node starts after the previous node finishes, so it can use the previous node's state updates.

### 2. Parallel Execution

Multiple nodes execute in the same graph step, allowing independent tasks to run concurrently. Their state updates are combined according to the reducers configured for each field.

### Example

#### Sequential Execution

`START → Researcher → Analyst → Writer → END`

- **Researcher:** Collects information.
- **Analyst:** Uses the researcher's results.
- **Writer:** Uses the analyst's output to generate the final report.

#### Parallel Execution

```text
          ┌→ Researcher ─┐
START ────┤              ├→ Writer → END
          └→ Analyst ────┘
```

- **Researcher:** Collects information.
- **Analyst:** Performs an independent analysis.
- **Writer:** Runs after both branches finish and combines their results.

### In Simple Words

**Sequential** means one task after another, while **parallel** means multiple independent tasks can run at the same time.


## 16. How do you implement branching and routing in a LangGraph workflow?


## Branching and Routing in LangGraph

In LangGraph, branching and routing are implemented using `add_conditional_edges()`. It allows the workflow to choose the next node based on the current state or a routing function's result.

### How It Works

1. **Define Nodes:** Create nodes for different tasks.
2. **Create a Routing Function:** Write a function that checks the current state and decides the next path.
3. **Add Conditional Edges:** Use `add_conditional_edges()` to map the routing result to the appropriate node.

### In Simple Words

Branching and routing allow LangGraph to choose which node should execute next based on the current state or a condition.


````python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

class AgentState(TypedDict):
    question: str
    category: str

def classify(state: AgentState):
    if "price" in state["question"].lower():
        return {"category": "pricing"}
    return {"category": "general"}

def router(state: AgentState):
    if state["category"] == "pricing":
        return "pricing_agent"
    return "general_agent"

def pricing_agent(state: AgentState):
    return {"category": "pricing"}

def general_agent(state: AgentState):
    return {"category": "general"}

graph = StateGraph(AgentState)

graph.add_node("classify", classify)
graph.add_node("pricing_agent", pricing_agent)
graph.add_node("general_agent", general_agent)

graph.add_edge(START, "classify")

graph.add_conditional_edges(
    "classify",
    router,
    {
        "pricing_agent": "pricing_agent",
        "general_agent": "general_agent"
    }
)

graph.add_edge("pricing_agent", END)
graph.add_edge("general_agent", END)

app = graph.compile()
````


### How It Works

- If the question contains `"price"`, the workflow routes to `pricing_agent`.
- Otherwise, it routes to `general_agent`.
- Both paths finish at `END`.

### In Simple Words

**Branching** creates different paths, and **routing** decides which path the workflow should follow based on the current state.


# 17. What is a supervisor-agent architecture, and how can you implement it using LangGraph?


## Supervisor-Agent Architecture in LangGraph

Supervisor-Agent Architecture is a multi-agent design where a supervisor agent manages multiple specialized agents. The supervisor decides which agent should execute a task and collects their results.

### How to Implement It Using LangGraph

1. **Create Shared State:** Store the task and results shared between agents.
2. **Define Specialized Agents:** Create agents such as `researcher` and `analyst` to perform specific tasks.
3. **Create a Supervisor Node:** Decide which agent should execute next.
4. **Add Conditional Edges:** Use `add_conditional_edges()` to route execution to the selected agent.
5. **Return to Supervisor:** After an agent finishes, return to the supervisor until the task is complete.

### In Simple Words

The supervisor acts like a team leader. It assigns tasks to specialized agents, reviews their results, and decides what should happen next until the task is completed.


````python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

class AgentState(TypedDict):
    task: str
    result: str
    next_agent: str

def supervisor(state: AgentState):
    if not state["result"]:
        return {"next_agent": "researcher"}
    return {"next_agent": "end"}

def researcher(state: AgentState):
    return {"result": "Research completed"}

def router(state: AgentState):
    return state["next_agent"]

graph = StateGraph(AgentState)

graph.add_node("supervisor", supervisor)
graph.add_node("researcher", researcher)

graph.add_edge(START, "supervisor")

graph.add_conditional_edges(
    "supervisor",
    router,
    {
        "researcher": "researcher",
        "end": END
    }
)

graph.add_edge("researcher", "supervisor")

app = graph.compile()

result = app.invoke({
    "task": "Research LangGraph",
    "result": "",
    "next_agent": ""
})

print(result)
````


### How It Works

- **Supervisor:** Checks the state and selects the researcher.
- **Researcher:** Performs the task and updates the result.
- **Return to Supervisor:** Control returns to the supervisor after the researcher finishes.
- **Complete Workflow:** The supervisor checks the state, sees that the task is complete, and ends the workflow.

### In Simple Words

The supervisor acts like a manager. It decides which agent should work next, while specialized agents perform their assigned tasks.


## 18. How do you implement tool calling in LangGraph?


## Tool Calling in LangGraph

Tool calling in LangGraph allows an AI agent to call external functions or tools, such as APIs, database queries, or search tools, when needed to complete a task.

You can implement it using an LLM with bound tools, `ToolNode`, and `tools_condition`.

### How It Works

1. **Define a Tool:** Create a tool using `@tool`.
2. **Bind the Tool:** Connect the tool to the LLM using `bind_tools()`.
3. **Create an Agent Node:** Call the LLM to decide whether a tool is needed.
4. **Execute the Tool:** Use `ToolNode` to execute the requested tools.
5. **Route Execution:** Use `tools_condition` to decide whether to execute a tool or finish the workflow.

### In Simple Words

Tool calling allows an AI agent to decide when it needs an external tool, execute that tool, and use the result to complete the task.


````python
from langchain_core.tools import tool
from langchain_groq import ChatGroq
from langgraph.graph import StateGraph, MessagesState, START, END
from langgraph.prebuilt import ToolNode, tools_condition

@tool
def add_numbers(a: int, b: int) -> int:
    """Add two numbers."""
    return a + b

tools = [add_numbers]

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
).bind_tools(tools)

def agent(state: MessagesState):
    response = llm.invoke(state["messages"])
    return {"messages": [response]}

graph = StateGraph(MessagesState)

graph.add_node("agent", agent)
graph.add_node("tools", ToolNode(tools))

graph.add_edge(START, "agent")

graph.add_conditional_edges(
    "agent",
    tools_condition,
    {
        "tools": "tools",
        END: END
    }
)

graph.add_edge("tools", "agent")

app = graph.compile()

result = app.invoke({
    "messages": [
        {"role": "user", "content": "What is 10 + 20?"}
    ]
})

print(result["messages"][-1].content)
````


### How It Works

- **Agent Node:** The LLM decides whether to call `add_numbers`.
- **ToolNode:** Executes the requested Python function.
- **tools_condition:** Routes to the tools node if a tool call is requested; otherwise, it ends the workflow.
- **Loop:** After the tool executes, its result is returned to the agent, which generates the final answer.

### In Simple Words

The LLM decides which tool to call, `ToolNode` executes it, and LangGraph manages the flow between the agent and the tool until the answer is ready.


## 19. What is the purpose of `ToolNode` and `tools_condition`?


### 1. ToolNode

`ToolNode` is a built-in LangGraph component that executes the tools requested by an LLM, such as Python functions, API calls, or database queries.

### 2. tools_condition

`tools_condition` checks the latest AI message and decides where the workflow should go next.

- If the AI message contains tool calls, it routes to the tools node.
- Otherwise, it routes to `END`.

### Example Flow

`START → Agent → tools_condition → ToolNode → Agent → END`

### In Simple Words

`tools_condition` decides whether a tool needs to run, and `ToolNode` executes that tool.


## 20.  How do you implement memory in LangGraph and its types?

Memory in LangGraph allows an AI agent to remember previous messages, user preferences, or information from earlier interactions.

### Types of Memory in LangGraph

### 1. Short-Term Memory

Stores information within a conversation or workflow thread. LangGraph uses state and checkpointing to maintain conversation history across multiple interactions.

### 2. Long-Term Memory

Stores information across different conversations or threads. LangGraph supports storing and retrieving memories using a Store.

### In Simple Words

- **Short-Term Memory:** Remembers information within a conversation.
- **Long-Term Memory:** Remembers information across different conversations.



````python
# Short Term
from langgraph.graph import StateGraph, MessagesState, START
from langgraph.checkpoint.memory import InMemorySaver

graph = StateGraph(MessagesState)

def chatbot(state):
    response = llm.invoke(state["messages"])
    return {"messages": [response]}

graph.add_node("chatbot", chatbot)
graph.add_edge(START, "chatbot")

app = graph.compile(checkpointer=InMemorySaver())

config = {"configurable": {"thread_id": "user123"}}

response = app.invoke(
    {"messages": [{"role": "user", "content": "My name is Ravi"}]},
    config=config
)

response = app.invoke(
    {"messages": [{"role": "user", "content": "What is my name?"}]},
    config=config
)
````


### How It Works

- **Checkpointer:** Saves the conversation state, including message history.
- **Thread ID:** Identifies a conversation thread.
- **Same Thread ID:** Allows the chatbot to access previous messages in that thread.
- **Different Thread ID:** Creates a separate conversation thread with its own state.

### In Simple Words

The checkpointer saves the conversation, and the `thread_id` helps LangGraph retrieve the previous conversation history when the user sends a new message.



### Long-Term Memory in LangGraph

LangGraph supports long-term memory using a Store, such as `InMemoryStore` or a persistent store. Memories are saved with a namespace and key, then retrieved when needed—even across different threads.

### Using PostgreSQL for Memory

PostgreSQL can persist memories across conversations and application restarts.

There are two common approaches:

- **PostgreSQL Checkpointer:** Saves workflow state and conversation history for short-term memory across interactions in the same thread.
- **PostgreSQL Store:** Saves long-term memories that can be retrieved across different threads or conversations.

LangGraph provides integrations through `langgraph-checkpoint-postgres`, including `PostgresSaver` for checkpointing and `PostgresStore` for long-term memory.

### In Simple Words

Use a **PostgreSQL checkpointer** to remember conversation state, and a **PostgreSQL store** to remember information across different conversations.


````python
from langgraph.store.postgres import PostgresStore

DB_URI = "postgresql://postgres:password@localhost:5432/langgraph"

with PostgresStore.from_conn_string(DB_URI) as store:
    store.setup()

    # Save long-term memory
    store.put(
        ("users", "user123"),
        "name",
        {"name": "Ravi"}
    )

    # Retrieve memory later
    memory = store.get(
        ("users", "user123"),
        "name"
    )

    print(memory.value)
````


### How It Works

- **`PostgresStore`:** Connects to PostgreSQL.
- **`store.put()`:** Saves the user's information.
- **`store.get()`:** Retrieves the saved information.
- **`("users", "user123")`:** The namespace, used to organize memories for a user.
- **`"name"`:** The key used to identify the memory.

### Important Notes

- Replace the connection string with your PostgreSQL credentials and database.
- PostgreSQL must be running.
- `store.setup()` creates the required tables.

### In Simple Words

This is long-term memory because the information remains stored in PostgreSQL and can be retrieved across different LangGraph threads and application restarts.


## 21. What are checkpointers, and why are they important?


## What Are Checkpointers in LangGraph?

**Checkpointers in LangGraph** save the graph's state at different steps of workflow execution. They allow the application to remember conversation history and resume execution when needed.

### Why Are They Important?

1. **Memory:** Preserve conversation history within a thread.
2. **Recovery:** Resume workflows after interruptions.
3. **Persistence:** Save state across application restarts when using persistent storage.
4. **Human-in-the-Loop:** Pause a workflow for human approval and resume it later.

### In Simple Words

Checkpointers act like save points in a game. They save the workflow's state so LangGraph can remember previous progress and resume execution when needed.


````python
from langgraph.graph import StateGraph, MessagesState, START
from langgraph.checkpoint.memory import InMemorySaver

graph = StateGraph(MessagesState)

def chatbot(state):
    response = llm.invoke(state["messages"])
    return {"messages": [response]}

graph.add_node("chatbot", chatbot)
graph.add_edge(START, "chatbot")

app = graph.compile(checkpointer=InMemorySaver())

config = {"configurable": {"thread_id": "user123"}}

result = app.invoke(
    {"messages": [{"role": "user", "content": "My name is Ravi"}]},
    config=config
)
````


### How It Works

- **`InMemorySaver()`:** Saves checkpoints in memory.
- **`thread_id`:** Identifies the conversation thread and helps retrieve its saved state.

### In Simple Words

A checkpointer works like a save point in a game. It saves the workflow's state so LangGraph can remember previous progress and continue when needed.

### Note

`InMemorySaver` loses its saved checkpoints when the application process ends. Use a persistent checkpointer, such as PostgreSQL, when you need state to survive application restarts.


## 22. What is the difference between `thread_id` and `checkpoint_id`?


### 1. `thread_id`

Identifies a conversation or workflow thread. LangGraph uses it to group checkpoints belonging to the same conversation.

### 2. `checkpoint_id`

Identifies a specific saved state (checkpoint) within that thread. It can be used to access or resume from a particular point in the workflow.

### Example

- **`thread_id = "user123"`** → Identifies the user's conversation.
- **`checkpoint_id = "abc456"`** → Identifies a particular saved state in that conversation.

### In Simple Words

`thread_id` identifies **which conversation**, while `checkpoint_id` identifies **which saved state within that conversation**.


## 23. How does human-in-the-loop functionality work in LangGraph?


## What Is Human-in-the-Loop (HITL) in LangGraph?

Human-in-the-loop (HITL) in LangGraph allows a workflow to pause and wait for human input or approval before continuing execution.

### How It Works

1. **Perform a Task:** The AI agent performs a task.
2. **Pause Execution:** LangGraph pauses execution using `interrupt()`.
3. **Human Review:** A human reviews the information and provides approval or feedback.
4. **Resume Workflow:** The workflow resumes using `Command(resume=...)`.
5. **Continue Execution:** The agent continues execution based on the human's response.

### In Simple Words

HITL allows an AI workflow to pause, ask a human for approval or feedback, and continue only after receiving the response.


````python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END
from langgraph.checkpoint.memory import InMemorySaver
from langgraph.types import interrupt, Command

class AgentState(TypedDict):
    task: str
    result: str

def approval_node(state: AgentState):
    decision = interrupt("Approve this task?")

    return {"result": decision}

graph = StateGraph(AgentState)
graph.add_node("approval", approval_node)

graph.add_edge(START, "approval")
graph.add_edge("approval", END)

app = graph.compile(checkpointer=InMemorySaver())

config = {"configurable": {"thread_id": "task123"}}

# Start workflow; it pauses for human input
result = app.invoke(
    {"task": "Deploy application", "result": ""},
    config=config
)

# Resume after human approval
result = app.invoke(
    Command(resume="Approved"),
    config=config
)

print(result["result"])
````


### How It Works

- **First Invocation:** The workflow pauses at `interrupt()` and waits for human input.
- **Human Approval:** The human supplies `"Approved"`.
- **Second Invocation:** The workflow resumes and continues execution using the human's response.

### Important

A checkpointer is required for interrupts to work correctly across invocations. In production, use persistent storage such as PostgreSQL.

### In Simple Words

Human-in-the-loop lets a human review, approve, or reject an AI agent's action before the workflow continues.


## 24. What are interrupts in LangGraph, and how do you resume execution after an interrupt?



Interrupts in LangGraph allow a workflow to pause execution at a specific point and wait for human input, approval, or feedback.

### How Do You Resume Execution?

1. **Pause Execution:** Use `interrupt()` to pause the workflow.
2. **Save State:** A checkpointer saves the workflow state.
3. **Human Input:** The human provides input or approval.
4. **Resume Execution:** Use `Command(resume=...)` with the same `thread_id` to resume execution.

### In Simple Words

Interrupts allow LangGraph to pause a workflow for human input and resume it from the saved state after receiving a response.


````python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END
from langgraph.checkpoint.memory import InMemorySaver
from langgraph.types import interrupt, Command

class AgentState(TypedDict):
    task: str
    approval: str

def approval_node(state: AgentState):
    decision = interrupt("Do you approve this task?")
    return {"approval": decision}

graph = StateGraph(AgentState)
graph.add_node("approval", approval_node)
graph.add_edge(START, "approval")
graph.add_edge("approval", END)

app = graph.compile(checkpointer=InMemorySaver())

config = {"configurable": {"thread_id": "task123"}}

# Start the workflow; it pauses at interrupt()
result = app.invoke(
    {"task": "Deploy application", "approval": ""},
    config=config
)

# Resume execution with human input
result = app.invoke(
    Command(resume="Approved"),
    config=config
)

print(result["approval"])
````


### Output

`Approved`

### Important Points

- **`interrupt()`:** Pauses the workflow and exposes a value for the caller to review.
- **`Command(resume=...)`:** Supplies the human's response to the paused workflow.
- **Same `thread_id`:** Required to resume the correct paused workflow.
- **Node Re-execution:** The node containing `interrupt()` starts executing again when resumed, so code before the interrupt may run again.

### In Simple Words

`interrupt()` pauses the workflow, and `Command(resume=...)` continues it with the human's response.


## 25.  What is the difference between `invoke()`, `stream()`, and `ainvoke()`?

### Execution Methods in LangGraph

### 1. `invoke()`

Executes the workflow and returns the final result after execution completes. It is synchronous, meaning it blocks until the result is ready.

### 2. `stream()`

Executes the workflow and yields updates as execution progresses. It is useful for monitoring node outputs or streaming intermediate results.

### 3. `ainvoke()`

Executes the workflow asynchronously and returns the final result. It uses `await` and is useful in asynchronous applications, such as FastAPI.

### In Simple Words

- **`invoke()`** — Runs the workflow and returns the final result.
- **`stream()`** — Runs the workflow and provides updates as it executes.
- **`ainvoke()`** — Runs the workflow asynchronously and returns the final result.



````python
# 1. Synchronous execution
result = app.invoke({"messages": ["Hello"]})

# 2. Stream intermediate updates
for update in app.stream({"messages": ["Hello"]}):
    print(update)

# 3. Asynchronous execution
result = await app.ainvoke({"messages": ["Hello"]})
````


### Key Differences

| Method | Execution | Returns |
|---|---|---|
| `invoke()` | Synchronous | Final result |
| `stream()` | Synchronous streaming | Iterator of updates or chunks |
| `ainvoke()` | Asynchronous | Final result when awaited |

### In Simple Words

- **`invoke()`** gives the final result.
- **`stream()`** provides updates during execution.
- **`ainvoke()`** gives the final result asynchronously.


## 26. How do you handle errors, retries, and failures in LangGraph workflows?



In LangGraph, errors and failures can be handled using exception handling, retries, timeouts, and checkpointing.

### 1. Error Handling

Use `try-except` blocks to catch exceptions and handle errors gracefully.

### 2. Retries

Use `RetryPolicy` to retry a node when a temporary failure occurs, such as an API timeout.

### 3. Timeouts

Set execution time limits to prevent a node from waiting indefinitely.

### 4. Checkpointing

Save workflow state so execution can recover after interruptions.

### 5. Fallbacks

Route execution to another node when a task fails or cannot be completed.

### In Simple Words

LangGraph helps handle failures by catching errors, retrying temporary failures, limiting execution time, saving progress, and switching to alternative paths when needed.


````python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END
from langgraph.types import RetryPolicy

class AgentState(TypedDict):
    result: str

def call_api(state: AgentState):
    try:
        # Simulate an API call
        result = "API call successful"
        return {"result": result}
    except Exception as e:
        return {"result": f"Error: {e}"}

graph = StateGraph(AgentState)

graph.add_node(
    "api",
    call_api,
    retry_policy=RetryPolicy(max_attempts=3)
)

graph.add_edge(START, "api")
graph.add_edge("api", END)

app = graph.compile()

result = app.invoke({"result": ""})
print(result)
````


### How It Works

- **`RetryPolicy(max_attempts=3)`:** Allows up to three attempts when the node raises a retryable exception.
- **`try-except`:** Catches exceptions and handles them. However, if you catch an exception and return normally, LangGraph will not retry that failure.
- **Exception Propagation:** For retries to work, the exception must propagate out of the node.

### In Simple Words

Use `try-except` for handling errors, `RetryPolicy` for retrying temporary failures, and checkpointing for saving workflow progress.


## 27. How do you implement persistence and recover a workflow after application interruption?

## What Is Persistence in LangGraph?

Persistence in LangGraph means saving workflow state using a checkpointer so the application can resume from a saved point after an interruption.

### How It Works

1. **Configure a Checkpointer:** Set up a checkpointer, such as PostgreSQL.
2. **Assign a Thread ID:** Give the workflow a unique `thread_id`.
3. **Save Checkpoints:** LangGraph saves checkpoints as the workflow executes.
4. **Restart the Application:** If the application stops, restart it and invoke the graph using the same `thread_id`.
5. **Restore State:** LangGraph can load the saved state and continue, depending on where execution stopped.

### In Simple Words

Persistence allows LangGraph to save workflow progress and recover the saved state after an interruption or application restart.



````python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END
from langgraph.checkpoint.postgres import PostgresSaver

class AgentState(TypedDict):
    task: str
    result: str

def process_task(state: AgentState):
    return {"result": "Task completed: " + state["task"]}

graph = StateGraph(AgentState)
graph.add_node("process", process_task)

graph.add_edge(START, "process")
graph.add_edge("process", END)

DB_URI = "postgresql://postgres:password@localhost:5432/langgraph"

with PostgresSaver.from_conn_string(DB_URI) as checkpointer:
    checkpointer.setup()

    app = graph.compile(checkpointer=checkpointer)

    config = {
        "configurable": {"thread_id": "task123"}
    }

    # Execute the workflow
    result = app.invoke(
        {"task": "Generate report", "result": ""},
        config=config
    )

    print(result)
````


### How Do You Recover After an Interruption?

After restarting the application, reconnect to PostgreSQL, create the graph with the same checkpointer configuration, and use the same `thread_id`.

- **Inspect Saved State:** Use `app.get_state(config)` to view the saved workflow state.
- **Resume an Interrupted Workflow:** Use `app.invoke(Command(resume=...), config)` to resume a workflow paused by an interrupt.
- **Continue from a Checkpoint:** LangGraph supports checkpoint-based execution using `Command(resume=...)` for interrupts or `Command(update=...)` and `Command(goto=...)` for other controlled continuation patterns.

### Important Note

Calling `app.invoke()` again with a fresh input and the same `thread_id` does not automatically mean a failed workflow will resume exactly where it stopped. Recovery behavior depends on the execution state and whether the workflow was interrupted, failed, or completed.

### In Simple Words

A PostgreSQL checkpointer saves workflow progress. After an interruption, LangGraph can recover the saved state using the same `thread_id`, with the appropriate resume method for that situation.

5. **Use Durable Workflows:** Combine LangGraph checkpointing with persistent operation records to recover safely after crashes.

Partial side effects can cause duplicate actions when a workflow resumes after a crash. Use idempotency, database records, and status checks to prevent duplicate operations.

## 29. What are the chucking stragies in LangGraph ?



Chunking strategies are techniques used to split large documents into smaller pieces before storing them in a vector database for RAG (Retrieval-Augmented Generation).

### Different Chunking Strategies

1. **Fixed-Size Chunking:** Splits text into chunks based on a fixed number of characters or tokens.

2. **Recursive Character Chunking:** Splits text using separators such as paragraphs, newlines, and sentences while trying to preserve context.

3. **Sentence-Based Chunking:** Splits text into individual sentences or groups of sentences.

4. **Paragraph-Based Chunking:** Splits documents based on paragraph boundaries.

5. **Semantic Chunking:** Splits text based on meaning, keeping related sentences together.

6. **Document Structure-Based Chunking:** Splits documents using headings, sections, tables, or chapters.

7. **Agentic Chunking:** Uses an LLM to decide where chunks should begin and end based on document context.

8. **Code-Based Chunking:** Splits source code based on functions, classes, and modules.

### In Simple Words

Chunking breaks large documents into smaller, meaningful pieces so a RAG application can retrieve relevant information more accurately.

### Interview Tip

- **Recursive character chunking** is a common starting point for RAG applications.
- **Semantic chunking** can be useful when preserving meaning is especially important.
- These are general RAG techniques, not features exclusive to LangGraph.
