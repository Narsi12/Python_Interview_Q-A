## 3. How does Python manage memory? Explain reference counting and garbage collection.

Python automatically manages memory using Reference Counting and Garbage Collection.

1. Reference Counting
Python tracks how many references point to an object.

````python
a = [1, 2, 3]  # One reference
b = a          # Two references

del a          # One reference remains
del b          # No named references remain

# When the reference count reaches zero, the object can be deallocated.
# del removes a reference or name; it does not directly delete the object from memory.
````

###### Garbage Collection (GC):
Garbage collection identifies and cleans up unreachable objects, especially circular references that reference counting alone cannot handle.

````python
a = []
a.append(a)  # Circular reference
del a
# Here, the list references itself. Even after del a, the circular reference remains. 
# Python's cyclic garbage collector can detect and collect this unreachable cycle.
````

## 4. What is the Global Interpreter Lock (GIL), and how does it affect multithreading?

##### What Is GIL?

- **GIL (Global Interpreter Lock):** A lock in the standard CPython interpreter that allows only one thread at a time to execute Python bytecode in a traditional GIL-enabled build.
- **Purpose:** It helps protect Python's internal data structures.
- **Limitation:** Multiple threads cannot execute Python bytecode simultaneously in the same interpreter under this model.
- **Example:** For CPU-heavy tasks, multiple threads may not improve performance because of the GIL.

##### Why Did Developers Create the GIL?
- The GIL was introduced in CPython to make memory management and interpreter operations safer and simpler when multiple threads exist.
- One important reason is that CPython uses reference counting to manage object memory.
- Remember our previous topic: when an object's reference count reaches zero, Python can deallocate it.
- Now imagine two threads modifying the same object's reference count at the same time. Without proper synchronization, they could interfere with each other and cause incorrect memory management.

##### The impact of the GIL depends on the type of task.
- **CPU-Bound Tasks:** These involve heavy calculations, such as mathematical operations or image processing. In traditional CPython builds, multiple threads cannot execute Python bytecode in parallel, so multithreading may not improve performance.
- **I/O-Bound Tasks:** These involve waiting for API responses, database queries, or file access. Multithreading is useful because a thread waiting for I/O can release the GIL, allowing another thread to run.

##### How Can We Overcome the GIL Limitation?

- **Multiprocessing:** Uses separate processes, each with its own Python interpreter, to run CPU-heavy tasks in parallel.
- **Multithreading:** Useful for I/O-bound tasks, where threads can work while other threads wait for responses.
- **`asyncio`:** Handles many I/O operations efficiently without creating a separate thread for each task.
- **Free-threaded CPython:** Allows Python threads to execute in parallel without the GIL in supported builds.



## 5. What is the difference between multithreading, multiprocessing, and asynchronous programming?

### Multithreading

- **Multiple Threads:** Uses multiple threads within a single process.
- **Best Use:** Suitable for I/O-bound tasks, such as API calls and database queries.
- **Shared Memory:** All threads share the same memory space.

````python
import threading
import time

def task(name):
    print(f"{name} started")
    time.sleep(2)
    print(f"{name} completed")

t1 = threading.Thread(target=task, args=("Task 1",))
t2 = threading.Thread(target=task, args=("Task 2",))

t1.start()
t2.start()

t1.join()
t2.join()

print("All tasks completed")
````

### Multiprocessing

- **Multiple Processes:** Uses multiple independent processes to run tasks.
- **Best Use:** Suitable for CPU-bound tasks, such as heavy calculations.
- **Separate Memory:** Each process has its own memory space.
- **Parallel Execution:** Processes can run in parallel across multiple CPU cores.

````python
from multiprocessing import Process

def task(name):
    print(f"{name} is running")

if __name__ == "__main__":
    p1 = Process(target=task, args=("Process 1",))
    p2 = Process(target=task, args=("Process 2",))

    p1.start()
    p2.start()

    p1.join()
    p2.join()

    print("All processes completed")
````

## 6. What is the difference between synchronous and asynchronous programming in Python?

##### Synchronous Programming
- **Sequential Execution:** Executes tasks one after another.
- **Waiting:** Each task waits for the previous task to finish.
- **Example:** Normal Python functions execute step by step.

````python
import time

def task1():
    time.sleep(2)
    print("Task 1 completed")

def task2():
    time.sleep(2)
    print("Task 2 completed")

task1()
task2()

print("All tasks completed")
````

### Asynchronous Programming

- **Non-Blocking Execution:** Allows a task to pause while waiting for I/O, so other tasks can run.
- **Best Use:** Useful for handling multiple API calls, database queries, and other I/O operations efficiently.
- **Keywords:** Uses `async` and `await` with tools such as `asyncio`.

- **`async`:** A keyword used to define a coroutine function in Python.
- **Execution:** Calling an `async` function creates a coroutine object; its body does not execute immediately.
- **coroutine:**A coroutine is a special function that can pause its execution and resume later without blocking other tasks from running.
In Python, we create a coroutine function using async and pause it using await.



````python
async def greet():
    print("Hello")

result = greet()
print(result)
# When you call greet(), Python creates a coroutine object. The print("Hello") statement runs when the coroutine is executed or awaited.
import asyncio

async def greet():
    print("Hello") # some times jupyternotebook through error try it on online compiler

asyncio.run(greet())
````

##### What is `await`?

- **`await`:** Pauses the current coroutine until an asynchronous operation completes.
- **Event Loop:** While the coroutine waits, the event loop can run other tasks.

````python
import asyncio
# run in online python compiler please
async def get_data():
    print("Fetching data...")
    await asyncio.sleep(2)
    print("Data received")

asyncio.run(get_data())
````

##### What is `asyncio`?

- **`asyncio`:** asyncio is a built-in Python library used to run multiple asynchronous tasks efficiently, especially when tasks spend time waiting for API responses, database queries, or network operations.

##### What does `asyncio` do?

- **Event Loop:** Manages and schedules asynchronous tasks.
- **Concurrency:** Allows multiple tasks to make progress while others wait.
- **`asyncio.gather()`:** Runs multiple awaitable operations concurrently.
- **`asyncio.sleep()`:** Pauses a coroutine without blocking the event loop.

````python
import asyncio

async def task1():
    print("Task 1 started")
    await asyncio.sleep(2)
    print("Task 1 completed")

async def task2():
    print("Task 2 started")
    await asyncio.sleep(2)
    print("Task 2 completed")

async def main():
    await asyncio.gather(task1(), task2())

asyncio.run(main())
````

##### What is `asyncio.run()`?

- **`asyncio.run()`:** Starts the event loop, runs an asynchronous function, and closes the event loop when the function finishes.
- **Purpose:** It is commonly used to run the main coroutine in a Python program.

````python
import asyncio

async def greet():
    print("Hello!")
    await asyncio.sleep(2)
    print("Welcome to Python")

asyncio.run(greet())
````

##### What is `asyncio.gather()`?

- **`asyncio.gather()`:** Runs multiple asynchronous tasks concurrently and collects their results.
- **Purpose:** It helps execute multiple tasks efficiently while other tasks are waiting.

````python
import asyncio

async def task1():
    print("Task 1 started")
    await asyncio.sleep(2)
    print("Task 1 completed")
    return "Result from Task 1"

async def task2():
    print("Task 2 started")
    await asyncio.sleep(2)
    print("Task 2 completed")
    return "Result from Task 2"

async def main():
    results = await asyncio.gather(task1(), task2())
    print(results)

asyncio.run(main())
````

## 7. Explain the LEGB rule and variable scope in Python. explain in simple english

#### LEGB Rule in Python

The **LEGB rule** defines the order in which Python searches for a variable.

- **L – Local:** Variables defined inside the current function.
- **E – Enclosing:** Variables defined inside an outer function.
- **G – Global:** Variables defined at the module level.
- **B – Built-in:** Python's predefined names, such as `print()`, `len()`, and `sum()`.


````python
x = "Global"  # G: Global scope

def outer():
    x = "Enclosing"  # E: Enclosing scope

    def inner():
        x = "Local"  # L: Local scope
        print(x)

    inner()
    print(x)

outer()
print(x)

print(len("Python"))  # B: Built-in scope
````

## 8. What are decorators? What are the practical use cases of decorators?
- A decorator is a function that adds extra functionality to another function without modifying its original code.

````python
def my_decorator(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")
    return wrapper

@my_decorator
def greet():
    print("Hello!")

greet()
````

````python
def calculator(func):
    def wrapper(*args):
        res = sum(args)*10
        return res
    return wrapper
@calculator
def vals_cal(a,b):
    return a+b
print(vals_cal(2,5))

# authentication function
def authentication(func):
    def wrapper(user):
        if user == 'admin':
            return func(user)
        else:
            return "Unathorized Access"
    return wrapper

@authentication
def user_access(user):
    return f"welcome {user}"
print(user_access("guhest"))
````

## 9. What are iterators, generators, and how are they different from normal functions?

#### What is an Iterator?

An **iterator** is an object that allows you to access elements one at a time.

Instead of getting all elements at once, an iterator gives you one element at a time.

#### `__iter__()`

- **`__iter__()`:** Returns an iterator object.
- Usually, we use `iter(numbers)` instead of calling `__iter__()` directly.

#### `__next__()`

- **`__next__()`:** Returns the next element from an iterator.
- Usually, we use `next(it)` instead of calling `__next__()` directly.

#### What is `StopIteration`?

- **`StopIteration`:** An exception raised when an iterator has no more elements.
- After returning all elements, the iterator raises `StopIteration` when you request another element.

````python
class MyNumbers:
    def __init__(self):
        self.numbers = [10, 20, 30]
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < len(self.numbers):
            value = self.numbers[self.index]
            self.index += 1
            return value
        else:
            raise StopIteration


obj = MyNumbers()
print(next(obj)) # or follow below loop to call one by one
for num in obj:
    print(num)
````

### Generators

A **generator** is a function that uses `yield` to produce values one at a time instead of returning all values at once.

1. **Iterator:** An object that implements `__iter__()` and `__next__()`.
2. **Generator:** A special type of iterator created using `yield` or a generator expression.

````python
def new_ele(n):
    for num in range(1, n+1):
        yield num  

x = new_ele(6)
print(next(x)) 
````

## 11. Explain shallow copy vs deep copy with their use cases.

##### 1. Shallow Copy
A **shallow copy** creates a new outer object, but nested objects are shared between the original and copied objects.

````python
import copy

original = [[1, 2], [3, 4]]
shallow = copy.copy(original)

shallow[0][0] = 100

print("Original:", original)
print("Shallow:", shallow)
````

##### 2. Deep Copy

A **deep copy** creates a completely independent copy of the original object, including its nested objects. Changes to the copied object won't affect the original.

````python
import copy

original = [[1, 2], [3, 4]]
deep = copy.deepcopy(original)

deep[0][0] = 100

print("Original:", original)
print("Deep Copy:", deep)
````

## 15.What is the difference between instance methods, class methods, and static methods?
### Instance Method
An **instance method** is a method inside a class that works with the data of a particular object.

- **`self`:** Refers to the current object.
- **Purpose:** Used to access or modify the object's attributes.
    

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):  # Instance method
        print(self.name, self.salary)

emp1 = Employee("Ravi", 50000)
emp2 = Employee("Kiran", 60000)

emp1.display()
emp2.display()
````

### Class Method

A **class method** is a method that works with the class itself rather than a specific object.

- **`cls`:** Refers to the current class.
- **Purpose:** Used to access or modify class-level attributes.
- **Limitation:** Cannot directly access instance-specific attributes.
- **Decorator:** Uses `@classmethod`.

````python
# Example 1
class Store:
    discount = 10  # Class attribute (default discount)
    
    @classmethod
    def set_discount(cls, new_discount):
        cls.discount = new_discount  # Modifying class attribute

# Changing discount using class method
Store.set_discount(15)
print(Store.discount)  # Output: 15

# Example 2
class Car:
    brand = "Toyota"  # Class attribute

    @classmethod
    def set_brand(cls, new_brand):
        cls.brand = new_brand  # Modifying class attribute

# Calling the class method
Car.set_brand("Honda")
print(Car.brand)  # Output: Honda


class Company:
    comapany_name = "Tech pod"
    def __init__(self,name):
        self.name = name
    @classmethod
    def modfiy(cls,new_name):
        cls.comapany_name = new_name
Company.modfiy("ND Tech")
print(Company.comapany_name)
name = Company("MicroChip")
print(name.comapany_name)
````

- `staticmethod`: is an independent method inside a class that does not depend on instance or class attributes. It behaves like a regular function but is grouped within a class for organizational purposes. It does not require self (instance) or cls (class).
  - **Example:** Calculating bank interest without depending on any specific bank account.

````python
# Example 1
class Bank:
    @staticmethod
    def calculate_interest(principal, rate, time):
        return (principal * rate * time) / 100  # Simple interest formula

# Calling static method without creating an instance
interest = Bank.calculate_interest(10000, 5, 2)
print(interest)  # Output: 1000.0

# Example 2
class MathUtils:
    @staticmethod
    def add(x, y):
        return x + y

# Calling the static method without creating an instance
result = MathUtils.add(5, 3)
print(result)  # Output: 8
````

### Hybrid Inheritance

**Hybrid inheritance** is a combination of two or more types of inheritance, such as single, multiple, hierarchical, or multilevel inheritance, in the same program.

- **Purpose:** Provides flexibility in designing class structures.

````python

# Base class
class Animal:
    def eat(self):
        print("Eats food")

# Intermediate class (inherits from Animal)
class Bird(Animal):
    def fly(self):
        print("Can fly")

# Another class (inherits from Animal)
class Fish(Animal):
    def swim(self):
        print("Can swim")

# Derived class (inherits from both Bird and Fish - Multiple Inheritance)
class Duck(Bird, Fish):
    def quack(self):
        print("Quacks")

# Creating an object of Duck
d = Duck()
d.eat()   # From Animal
d.fly()   # From Bird
d.swim()  # From Fish
d.quack() # From Duck
````

## Polymorphism

**Polymorphism** is the ability of an object or method to take different forms.

In Python, polymorphism can be achieved through method overriding and method overloading.

### 1. Method Overriding

**Method overriding** occurs when a child class defines a method with the same name as a method in the parent class but provides different functionality.

### 2. Method Overloading

**Method overloading** means defining multiple methods with the same name but different parameters.

**Note:** Python does not support traditional method overloading directly. Similar behavior can be achieved using default arguments or `*args`.

````python
# Method Overriding
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):  # Method overriding
        print("Dog barks")

class Cat(Animal):
    def sound(self):  # Method overriding
        print("Cat meows")


animal = Animal()
dog = Dog()
cat = Cat()

animal.sound()
dog.sound()
cat.sound()
````

````python
class A:
    def show(self):
        print("A")

class B(A):
    pass

class C(B):
    pass

class D(B, C):
    pass

print(D.mro())
print(D.__mro__)
````

## What is `__new__()`?

`__new__()` is a special method responsible for creating a new object and returning it.

- **`cls`:** Refers to the class being instantiated.
- **Execution:** Runs before `__init__()`.

- **Use Case:** Useful for controlling object creation, such as implementing a Singleton or creating immutable objects like `str` and `tuple`.

## 19. How does exception propagation work in Python? What are the best practices for custom exceptions?
**Exception Propagation:** When an error occurs inside a function and the function doesn't handle it, the error moves to the calling function until a matching `except` block handles it.

## 20. What are mutable default arguments, and why can they cause unexpected behavior?
**Mutable Default Arguments:** Mutable objects, such as lists or dictionaries, used as default values in function parameters.

Python creates the default object only once when the function is defined, so changes to that object can persist across multiple function calls.

````python
def add_item(item, items=[]):
    items.append(item)
    return items

print(add_item("Apple"))
print(add_item("Banana"))
print(add_item("Orange"))
````

`Why does this happen?`
- The default list [] is created only once.
- Every call reuses the same list.
- New items are added to the existing list instead of starting with an empty list.

`Correct approach`
- Use None as the default value and create a new list inside the function.

````python
def add_item(item, items=None):
    if items is None:
        items = []

    items.append(item)
    return items

print(add_item("Apple"))
print(add_item("Banana"))
print(add_item("Orange"))
````

## 23. What are metaclasses, and in which scenarios would you use them?

**Metaclass:** A metaclass is a class that creates other classes. In Python, the default metaclass is `type`.

Normally, a class creates objects, while a metaclass controls how a class itself is created.

````python
class Student:
    pass

print(type(Student))  # <class 'type'>
#Here, type is the metaclass that created the Student class.
````

````python
# Custom Metaclass Example
class MyMeta(type):
    def __new__(mcs, name, bases, attrs):
        print(f"Creating class: {name}")
        return super().__new__(mcs, name, bases, attrs)

class Student(metaclass=MyMeta):
    pass
````

`mcs:` Similar to how `self` refers to an object, mcs refers to the metaclass. The name `mcs` is a convention; you could call it something else.

`name:` The name of the class Python is creating.

`bases:` The parent classes inherited by the new class. For example, if Student inherits from Person, it contains Person.

`attrs:` The class's namespace: its attributes and methods defined in the class body.

## 24. What is duck typing, and how does Python support polymorphism?
**Duck Typing:** Duck typing means Python focuses on what an object can do, rather than its actual class or type.

The idea is: “If it walks like a duck and quacks like a duck, treat it like a duck.”

3. `make_sound()` doesn't check the object's class.
4. It simply calls `animal.sound()`.
5. Python executes the appropriate method for each object.

**Polymorphism** means the same interface or method call can produce different behavior depending on the object.

In the example above, `animal.speak()` produces different results for `Dog` and `Cat`.

Python supports polymorphism through:

## 30. What are the SOLID principles in object-oriented programming, and how does each principle help in designing maintainable, scalable, and reusable Python applications?

**SOLID:** A set of five principles that help developers write maintainable, reusable, and scalable code.

### 1. S — Single Responsibility Principle (SRP)

A class should have only one responsibility.

````python
class Report:
    def generate(self):
        print("Generating report")

class ReportSaver:
    def save(self, report):
        print("Saving report")

# Benefit: Changes to report generation don't affect saving logic.
````

### 2. O — Open/Closed Principle (OCP)

**Definition:** Code should be open for extension but closed for modification.

This means we should be able to add new features without changing existing code.

````python
# ❌ Without OCP
def calculate_area(shape, value):
    if shape == "circle":
        return 3.14 * value * value
    elif shape == "square":
        return value * value

print(calculate_area("circle", 5))
print(calculate_area("square", 4))

# If we add a new shape, like a rectangle, we must modify the existing function.

# ✅ With OCP
class Circle:
    def area(self):
        return 3.14 * 5 * 5

class Square:
    def area(self):
        return 4 * 4

def calculate_area(shape):
    return shape.area()

print(calculate_area(Circle()))
print(calculate_area(Square()))

````

### How Does OCP Work Here?

- **Flexibility:** `calculate_area()` works with any object that provides an `area()` method.
- **Extension:** To add a new shape, create a new class with its own `area()` method.
- **No Modification:** You don't need to modify `calculate_area()`.

### 3.L — Liskov Substitution Principle (LSP)

**Simple Definition:** A child class should be usable wherever its parent class is expected, without breaking the program.

````python
class Bird:
    def eat(self):
        print("Bird is eating")

class Sparrow(Bird):
    def fly(self):
        print("Sparrow is flying")

def make_bird_eat(bird):
    bird.eat()

sparrow = Sparrow()
make_bird_eat(sparrow)
````

**LSP (Liskov Substitution Principle)** means a child class should not break the expected logic or behavior of its parent class.

For example:

- **Parent Class:** `Bird` provides an `eat()` method.
- **Child Class:** `Sparrow` inherits from `Bird` and can eat normally.
- **Substitution:** Wherever a `Bird` object is expected, passing a `Sparrow` should work correctly.

**Remember:** Child classes should extend the parent class without breaking its expected behavior.

### 4. Interface Segregation Principle (ISP)

**Simple Definition:** A class should not be forced to implement methods that it does not need.

**Example:** Imagine a `Bird` class has two methods: `fly()` and `swim()`.

- **Sparrow:** Can fly but cannot swim. Forcing `Sparrow` to implement `swim()` is unnecessary.

**Purpose:** Create small, separate interfaces so each class only uses the methods it actually needs.

**Remember:** Don't force a class to implement unnecessary methods.

````python
# ❌ Without ISP
class Bird:
    def fly(self):
        print("Flying")

    def swim(self):
        print("Swimming")


class Sparrow(Bird):
    def fly(self):
        print("Sparrow is flying")

    def swim(self):
        raise NotImplementedError("Sparrow cannot swim")

# ✅ With ISP
class FlyingBird:
    def fly(self):
        pass

class SwimmingBird:
    def swim(self):
        pass

class Sparrow(FlyingBird):
    def fly(self):
        print("Sparrow is flying")

class Duck(FlyingBird, SwimmingBird):
    def fly(self):
        print("Duck is flying")

    def swim(self):
        print("Duck is swimming")


sparrow = Sparrow()
sparrow.fly()

duck = Duck()
duck.swim()
````

### D — Dependency Inversion Principle (DIP)

**Simple Definition:** High-level business logic should not depend directly on a specific tool or implementation. Both should depend on a common interface.

**Simple Example:** Imagine your application sends notifications.

If your code directly depends on Gmail, switching to Outlook requires changing your business logic. With DIP, your code depends on a common notification interface, so you can switch implementations more easily.

**Remember:** Depend on an abstraction, not on a specific implementation.

````python
# ❌ Without DIP
class Email:
    def send(self, message):
        print("Sending email:", message)

class Notification:
    def __init__(self):
        self.email = Email()  # Direct dependency

    def notify(self, message):
        self.email.send(message)

n = Notification()
n.notify("Hello!")
# Here, Notification directly depends on Email. Switching to SMS requires changing the Notification class.

# ✅ With DIP

class Email:
    def send(self, message):
        print("Sending email:", message)

class SMS:
    def send(self, message):
        print("Sending SMS:", message)

class Notification:
    def __init__(self, service):
        self.service = service

    def notify(self, message):
        self.service.send(message)

n1 = Notification(Email())
n1.notify("Hello!")

n2 = Notification(SMS())
n2.notify("Hello!")
````

1. **Business Logic:** `Notification` contains the business logic.
2. **Implementations:** `Email` and `SMS` are different implementations.
3. **Dependency Injection:** `Notification` receives an implementation through its constructor instead of creating it itself.
4. **Flexibility:** You can switch between email and SMS without modifying the `Notification` class.
